## Question 1: Lists

```python
# Roll number and digit extraction
roll_number = "1024170372"
digits = [int(d) for d in roll_number]
print("Digits of roll number:", digits)

# Multiply each digit by 10 to form list L
L = [d * 10 for d in digits]
print("List L:", L)
```

**i. Print L**

```python
print(L)
```

**ii. Add two numbers — one using append(), one using insert()**

```python
# append() adds an element at the END of the list
L.append(99)
print("After append(99):", L)   # 99 gets added at the last position

# insert() adds an element at a SPECIFIC index/position
L.insert(3, 55)
print("After insert(3, 55):", L)  # 55 is inserted at index 3, shifting later elements right
```

**iii. Remove two elements — one using remove(), one using pop()**

```python
# remove() deletes the first occurrence of a given VALUE
L.remove(99)
print("After remove(99):", L)   # removes value 99 wherever it first occurs

# pop() removes an element at a given INDEX (or last element if no index given)
popped_value = L.pop()
print(f"After pop() (removed {popped_value}):", L)  # removes and returns the last element
```

**iv. Sort L ascending, then descending**

```python
L.sort()
print("Ascending order:", L)

L.sort(reverse=True)
print("Descending order:", L)
```

**v. Slicing — first three and last three elements**

```python
print("First three elements:", L[:3])
print("Last three elements:", L[-3:])
```

**vi. List comprehension — elements greater than the average of L**

```python
average_L = sum(L) / len(L)
above_average = [x for x in L if x > average_L]
print("Average of L:", average_L)
print("Elements greater than average:", above_average)
```

## Question 2: Tuples

```python
# Tuple of 8 scores using the first 8 values from list L in Q1 (before the append/insert/sort edits)
scores = tuple(d * 10 for d in digits[:8])
print("scores:", scores)
```

**i. Highest score & its index, lowest score & its frequency**

```python
highest_score = max(scores)
highest_index = scores.index(highest_score)
lowest_score = min(scores)
lowest_count = scores.count(lowest_score)

print(f"Highest score: {highest_score} at index {highest_index}")
print(f"Lowest score: {lowest_score}, appears {lowest_count} time(s)")
```

**ii. Reverse the tuple (as a list)**

Tuples are immutable, so there is no in-place `.reverse()` method for them (that would require modifying
the tuple after creation, which Python does not allow). Instead we convert to a list, reverse it, and can
optionally convert back.

```python
reversed_scores = list(reversed(scores))
print("Reversed scores (as list):", reversed_scores)
```

**iii. Ask user for a score and find its first occurrence index**

```python
try:
    user_score = int(input("Enter a score to search for in the tuple: "))
except Exception:
    user_score = scores[2]
    print(f"(No interactive input available - using sample value {user_score} for demonstration)")

if user_score in scores:
    print(f"{user_score} found at index {scores.index(user_score)}")
else:
    print(f"{user_score} is not present in the tuple")
```

**iv. Attempt to change an element directly**

```python
try:
    scores[0] = 100
except TypeError as e:
    print("Error raised:", e)
    print("Explanation: tuples are immutable, so item assignment is not allowed;",
          "a list would allow this because lists are mutable.")
```

**v. Unpack the tuple: first, second, and the rest**

```python
first_score, second_score, *remaining_scores = scores
print("First score:", first_score)
print("Second score:", second_score)
print("Remaining scores:", remaining_scores)
```

## Question 3: Random Numbers & List Operations

```python
import random

random.seed(1024170372)  # seed = own roll number, for a reproducible-but-unique list

random_numbers = [random.randint(100, 900) for _ in range(100)]
print("Generated list of 100 random numbers:")
print(random_numbers)
```

**ii. Count of odd numbers**

```python
odd_numbers = [n for n in random_numbers if n % 2 != 0]
print("Count of odd numbers:", len(odd_numbers))
```

**iii. Count of even numbers**

```python
even_numbers = [n for n in random_numbers if n % 2 == 0]
print("Count of even numbers:", len(even_numbers))
```

**iv. Count and list of prime numbers**

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

prime_numbers = [n for n in random_numbers if is_prime(n)]
print("Count of prime numbers:", len(prime_numbers))
print("Prime numbers found:", prime_numbers)
```

**v. Most frequently occurring number**

```python
most_common = max(set(random_numbers), key=random_numbers.count)
print(f"Most frequent number: {most_common}, occurs {random_numbers.count(most_common)} time(s)")
```

## Question 4: Sets

```python
# Use the 8 original digits (before multiplying by 10) from Q1
digits_8 = digits[:8]
print("First 8 digits:", digits_8)

A = {d * 7 for d in digits_8}
B = {d * 9 for d in digits_8}

print("Set A (digit x 7):", A)
print("Set B (digit x 9):", B)
```

**vi. Union of A and B**

```python
union_AB = A.union(B)
print("A union B:", union_AB)
```

**vii. Intersection of A and B**

```python
intersection_AB = A.intersection(B)
print("A intersection B:", intersection_AB)
```

**viii. A − B and B − A**

```python
A_minus_B = A.difference(B)
B_minus_A = B.difference(A)
print("A - B:", A_minus_B)
print("B - A:", B_minus_A)

# difference() gives elements in ONE set but not the other (direction matters: A-B != B-A in general)
# symmetric_difference() gives elements that are in EITHER set but NOT in both, combined into a single set
# i.e. symmetric_difference(A, B) == (A - B) union (B - A)
```

**ix. Symmetric difference of A and B**

```python
symmetric_diff = A.symmetric_difference(B)
print("Symmetric difference:", symmetric_diff)
```

**x. Subset / superset checks**

```python
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))
```

**xi. Ask user for value X and discard() it from A**

```python
try:
    X = int(input("Enter a value X to remove from set A: "))
except Exception:
    X = next(iter(A))
    print(f"(No interactive input available - using sample value {X} for demonstration)")

print("Set A before discard:", A)
A.discard(X)  # discard() does NOT raise an error if X is not present, unlike remove()
print("Set A after discard:", A)
print("discard() is safer than remove() here because remove() would raise a KeyError",
      "if the value is not found, while discard() simply does nothing.")
```

## Question 5: Dictionaries

```python
my_dict = {
    "name": "Ashish Bhagat",
    "roll_no": "1024170372",
    "branch": "COPC",
    "age": 20,
    "city": "Jammu"
}
print(my_dict)
```

**i. Rename key 'city' to 'location' (using pop())**

```python
my_dict["location"] = my_dict.pop("city")
print(my_dict)
```

**ii. Add a new key 'cgpa'**

```python
my_dict["cgpa"] = 7.40
print(my_dict)
```

**iii. Increase 'age' by 1**

```python
my_dict["age"] += 1
print(my_dict)
```

**iv. Delete 'branch' key using pop() and del in two separate copies**

```python
dict_copy_1 = my_dict.copy()
dict_copy_2 = my_dict.copy()

popped_branch = dict_copy_1.pop("branch")
print("dict_copy_1 after pop('branch'):", dict_copy_1)
print("pop() returned the removed value:", popped_branch)

del dict_copy_2["branch"]
print("dict_copy_2 after del:", dict_copy_2)

# pop(key) removes the key AND returns its value, so it can be captured in a variable;
# del key just deletes the entry and returns nothing.
```

**v. Iterate using .items()**

```python
for key, value in my_dict.items():
    print(f"{key} → {value}")
```

**vi. Safely check for a key that doesn't exist ('email')**

```python
if "email" in my_dict:
    print(my_dict["email"])
else:
    print("Key 'email' not found — no email address on record.")
```

**vii. Merge with friend_dict using {**dict1, **dict2}**

```python
friend_dict = {
    "name": "Rohan Sharma",
    "roll_no": "1024170399",
    "branch": "CSE",
    "age": 21,
    "city": "Delhi"
}

merged_dict = {**my_dict, **friend_dict}
print("Merged dictionary:", merged_dict)
print("When both dictionaries share a key, the value from the dictionary listed SECOND",
      "(friend_dict here) overwrites the first one's value.")
```

**viii. Dictionary comprehension — keep only string-valued pairs**

```python
string_only_dict = {k: v for k, v in my_dict.items() if isinstance(v, str)}
print("Only string-valued key-value pairs:", string_only_dict)
```
