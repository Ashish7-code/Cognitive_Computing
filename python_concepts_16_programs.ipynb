## 1. Syntax Basics

Python syntax highlights: variables need no type declaration, indentation defines blocks (no braces), `#` starts a comment, and `type()` reveals the underlying type of any object.

```python
# Variables — no type declaration needed
name = "Animesh"          # str
age = 20                  # int
gpa = 8.7                 # float
is_student = True         # bool

# Comments start with '#'
print("Name:", name, "| Type:", type(name))
print("Age:", age, "| Type:", type(age))
print("GPA:", gpa, "| Type:", type(gpa))
print("Is student:", is_student, "| Type:", type(is_student))

# Indentation defines a block (here, the body of the if statement)
if age >= 18:
    print(f"{name} is an adult.")
else:
    print(f"{name} is a minor.")

# Multiple assignment in one line
x, y, z = 1, 2, 3
print("x, y, z =", x, y, z)
```

## 2. Loops — `for`

`for` loops iterate over any *iterable* (ranges, lists, strings, etc.). `enumerate()` gives index + value together.

```python
# Looping over a range
print("Squares from 1 to 5:")
for i in range(1, 6):
    print(f"  {i}^2 = {i**2}")

# Looping over a list with enumerate
fruits = ["apple", "banana", "cherry"]
print("\nFruit list:")
for index, fruit in enumerate(fruits, start=1):
    print(f"  {index}. {fruit}")

# Looping over a string
print("\nCharacters in 'Python':")
for ch in "Python":
    print(" ", ch)
```

## 3. Loops — `while`

`while` loops run until a condition becomes false. `break` exits early, `continue` skips to the next iteration.

```python
# Countdown using while
count = 5
print("Countdown:")
while count > 0:
    print(" ", count)
    count -= 1
print("  Liftoff!")

# while with break and continue
print("\nPrinting odd numbers below 10, stopping if we hit 7:")
n = 0
while n < 10:
    n += 1
    if n % 2 == 0:
        continue        # skip even numbers
    if n == 7:
        print("  Reached 7, stopping.")
        break
    print(" ", n)
```

## 4. Conditionals — `if / elif / else`

A simple grading system demonstrates branching logic.

```python
def grade(score):
    if score >= 90:
        return "A"
    elif score >= 75:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 40:
        return "D"
    else:
        return "F"

scores = [95, 82, 67, 41, 30]
for s in scores:
    print(f"Score {s:>3} -> Grade {grade(s)}")
```

## 5. Data Structures — List

Lists are ordered, mutable collections. Common operations: append, insert, remove, slicing, sorting, and comprehensions.

```python
numbers = [5, 3, 8, 1, 9, 2]
print("Original list:", numbers)

numbers.append(12)          # add to end
numbers.insert(0, 100)      # insert at position
numbers.remove(3)           # remove by value
print("After modifications:", numbers)

print("Sliced [1:4]:", numbers[1:4])
print("Sorted:", sorted(numbers))
print("Reversed:", numbers[::-1])

# List comprehension: squares of even numbers
squares_of_evens = [n**2 for n in numbers if n % 2 == 0]
print("Squares of even numbers:", squares_of_evens)
```

## 6. Data Structures — Dictionary

Dictionaries store key-value pairs. They support fast lookup, iteration over keys/values/items, and comprehensions.

```python
student = {
    "name": "Animesh",
    "roll_no": 101,
    "courses": ["Discrete Math", "Numerical Linear Algebra", "Mathematics 2"]
}
print("Student record:", student)

# Access, update, add
student["gpa"] = 8.7
student["roll_no"] = 102
print("\nUpdated record:", student)

print("\nKeys:", list(student.keys()))
print("Values:", list(student.values()))

print("\nIterating items:")
for key, value in student.items():
    print(f"  {key}: {value}")

# Dictionary comprehension
squares_dict = {n: n**2 for n in range(1, 6)}
print("\nSquares dictionary:", squares_dict)
```

## 7. Data Structures — Tuple & Set

Tuples are ordered and **immutable**. Sets are unordered collections of unique elements, useful for membership tests and set algebra.

```python
# Tuple — immutable, good for fixed records
point = (3, 4)
x, y = point   # unpacking
print("Point:", point, "-> x =", x, ", y =", y)

try:
    point[0] = 10   # this will fail, tuples can't be modified
except TypeError as e:
    print("Cannot modify tuple:", e)

# Set — unique, unordered elements
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print("\nSet a:", a)
print("Set b:", b)
print("Union:", a | b)
print("Intersection:", a & b)
print("Difference (a - b):", a - b)

# Removing duplicates from a list using a set
raw = [1, 2, 2, 3, 3, 3, 4]
print("\nDeduplicated list:", list(set(raw)))
```

## 8. Strings — Manipulation

Strings are immutable sequences. Slicing, case methods, splitting/joining, and searching are core operations.

```python
text = "  Python is Fun and Powerful  "
print(f"Original: '{text}'")

stripped = text.strip()
print(f"Stripped: '{stripped}'")
print("Upper:", stripped.upper())
print("Lower:", stripped.lower())
print("Title:", stripped.title())

words = stripped.split()
print("\nWords:", words)
print("Joined with '-':", "-".join(words))

print("\nSlicing 'Python'[0:3]:", stripped[0:6])
print("Reversed string:", stripped[::-1])
print("Does it contain 'Fun'?", "Fun" in stripped)
print("Replace 'Fun' with 'Awesome':", stripped.replace("Fun", "Awesome"))
```

## 9. Strings — Formatting

Three common ways to format strings: f-strings (modern, preferred), `.format()`, and the older `%` operator.

```python
name = "Animesh"
score = 87.456
total = 100

# f-strings (Python 3.6+)
print(f"f-string: {name} scored {score:.2f} out of {total} ({score/total:.1%})")

# .format() method
print("format(): {} scored {:.2f} out of {} ({:.1%})".format(name, score, total, score/total))

# % operator (legacy style)
print("%% operator: %s scored %.2f out of %d" % (name, score, total))

# Alignment and padding
for item, price in [("Notebook", 45), ("Pen", 10), ("Eraser", 5)]:
    print(f"{item:<10}Rs.{price:>5}")
```

## 10. File Handling

Writing to, appending to, and reading from a text file using the `with` context manager (which auto-closes the file).

```python
filename = "sample_notes.txt"

# Write mode — creates/overwrites the file
with open(filename, "w") as f:
    f.write("Line 1: Learning Python file handling.\n")
    f.write("Line 2: The 'with' statement auto-closes files.\n")

# Append mode — adds to existing content
with open(filename, "a") as f:
    f.write("Line 3: Appended later.\n")

# Read mode — read the whole file
with open(filename, "r") as f:
    content = f.read()
print("Full file content:")
print(content)

# Read line by line
print("Reading line by line:")
with open(filename, "r") as f:
    for i, line in enumerate(f, 1):
        print(f"  {i}: {line.strip()}")

import os
os.remove(filename)  # cleanup
print("\n(cleaned up sample_notes.txt)")
```

## 11. Exception Handling — Basics

`try/except/else/finally` handles runtime errors gracefully instead of crashing the program.

```python
def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print(f"  Error: cannot divide {a} by zero.")
        return None
    except TypeError:
        print(f"  Error: invalid types for division ({a!r}, {b!r}).")
        return None
    else:
        print(f"  Success: {a} / {b} = {result}")
        return result
    finally:
        print("  (division attempt finished)\n")

safe_divide(10, 2)
safe_divide(10, 0)
safe_divide(10, "two")
```

## 12. Exception Handling — Custom Exceptions

Custom exception classes let you define domain-specific errors by subclassing `Exception`.

```python
class InsufficientBalanceError(Exception):
    """Raised when a withdrawal exceeds the available balance."""
    def __init__(self, balance, amount):
        self.balance = balance
        self.amount = amount
        super().__init__(f"Cannot withdraw Rs.{amount}, balance is only Rs.{balance}")

def withdraw(balance, amount):
    if amount > balance:
        raise InsufficientBalanceError(balance, amount)
    return balance - amount

balance = 500
for amount in [200, 1000]:
    try:
        balance = withdraw(balance, amount)
        print(f"Withdrew Rs.{amount}, new balance: Rs.{balance}")
    except InsufficientBalanceError as e:
        print("Transaction failed:", e)
```

## 13. Random Numbers

The `random` module generates pseudo-random numbers, choices, and shuffles. `random.seed()` makes output reproducible.

```python
import random

random.seed(42)   # reproducible output for this demo

print("Random int between 1 and 100:", random.randint(1, 100))
print("Random float between 0 and 1:", random.random())
print("Random float between 5 and 10:", random.uniform(5, 10))

deck = ["A", "K", "Q", "J", "10"]
print("\nRandom choice from deck:", random.choice(deck))
print("Random sample of 3 (no repeats):", random.sample(deck, 3))

random.shuffle(deck)
print("Shuffled deck:", deck)
```

## 14. Command Line Arguments

Since a notebook has no real command line, this program writes a standalone `.py` script that reads `sys.argv`
and `argparse`, then runs it via `subprocess` with sample arguments to show real CLI behaviour.

```python
script = '''
import sys
import argparse

print("Raw sys.argv:", sys.argv)

parser = argparse.ArgumentParser(description="Greet a user N times.")
parser.add_argument("name", help="Name of the person to greet")
parser.add_argument("--times", type=int, default=1, help="Number of greetings")
args = parser.parse_args()

for _ in range(args.times):
    print(f"Hello, {args.name}!")
'''

with open("cli_demo.py", "w") as f:
    f.write(script)

import subprocess
result = subprocess.run(
    ["python3", "cli_demo.py", "Animesh", "--times", "3"],
    capture_output=True, text=True
)
print("--- Script output ---")
print(result.stdout)

import os
os.remove("cli_demo.py")  # cleanup
```

## 15. Libraries — `math` & `datetime`

Python's standard library ships many ready-made modules. `math` covers numeric functions; `datetime` covers dates and times.

```python
import math
from datetime import datetime, timedelta

print("math.sqrt(64):", math.sqrt(64))
print("math.factorial(5):", math.factorial(5))
print("math.pi:", math.pi)
print("math.ceil(4.2), math.floor(4.8):", math.ceil(4.2), math.floor(4.8))

now = datetime.now()
print("\nCurrent datetime:", now.strftime("%Y-%m-%d %H:%M:%S"))
future = now + timedelta(days=30)
print("30 days from now:", future.strftime("%Y-%m-%d"))
print("Day of week today:", now.strftime("%A"))
```

## 16. Libraries — `collections` & `itertools`

`collections` provides specialized container types (`Counter`, `defaultdict`); `itertools` provides fast, memory-efficient iterators (`permutations`, `combinations`).

```python
from collections import Counter, defaultdict
from itertools import permutations, combinations

words = "the quick brown fox jumps over the lazy dog the fox runs".split()
word_counts = Counter(words)
print("Word frequency:", word_counts)
print("Most common 2 words:", word_counts.most_common(2))

groups = defaultdict(list)
for w in words:
    groups[len(w)].append(w)
print("\nWords grouped by length:")
for length, ws in sorted(groups.items()):
    print(f"  {length} letters: {ws}")

letters = ["A", "B", "C"]
print("\nPermutations of", letters, ":", list(permutations(letters, 2)))
print("Combinations of", letters, "(choose 2):", list(combinations(letters, 2)))
```
